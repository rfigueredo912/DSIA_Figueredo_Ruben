```
# Ejercicio 1 – Variables y operaciones

nombre = "Rubén Darío Figueredo"
edad = 36 
ciudad = "CABA"

print(f"Hola, mi nombre es {nombre}, tengo {edad} años y vivo en {ciudad}.")

# Variables numericas y operaciones
a = 15
b = 4

print("Suma:", a + b)
print("Resta:", a - b)
print("Multiplicacion:", a * b)
print("División:", a / b)

# Ejercicio 2 – Condicionales

edad_ingresada = int(input("Ingresa tu edad: "))

if edad_ingresada >= 18:
    print("Eres mayor de edad")
else:
    print("Eres menor de edad")

# Ejercicio 3 – Listas y bucles

numeros = [10, 20, 30, 40, 50]

print("Elementos de la lista:")
for numero in numeros:
    print(numero)

# Calculo de suma y promedio
suma_total = sum(numeros)
promedio = suma_total / len(numeros)

print("\nSuma total:", suma_total)
print("Promedio:", promedio)    
```
