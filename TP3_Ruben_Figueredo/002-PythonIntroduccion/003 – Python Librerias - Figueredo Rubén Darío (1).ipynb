```
# Ejercicio 2 – Dataset simple con Pandas

# Crear DataFrame
data = {
    'estudiante': ['Ana', 'Juan', 'Carla', 'Pedro', 'Lucia'],
    'nota': [8, 6, 9, 5, 7]
}
df = pd.DataFrame(data)

# Mostrar operaciones
print("DataFrame de estudiantes:")
print(df)
print("\n--- Estadísticas de las notas ---")
print("Promedio de las notas:", df['nota'].mean())
print("Nota máxima:", df['nota'].max())
print("Nota mínima:", df['nota'].min())
```
